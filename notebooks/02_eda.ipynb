# Football-specific exploratory data analysis
Run notebook 01 first. These plots use only columns present in the source; xG and xAG are unavailable.

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('data/processed/players_cleaned.csv')
display(data[['Min','Gls','Ast','Age']].describe(), data['Comp'].value_counts(), data['Position_Category'].value_counts())
for field in ['Gls', 'Ast', 'Min', 'Age']:
    data[field].hist(bins=25); plt.title(f'{field} distribution'); plt.xlabel(field); plt.show()
for x, y in [('Sh','Gls'), ('SoT','Gls'), ('Crs','Ast'), ('Age','Gls_per90')]:
    data.plot.scatter(x=x, y=y, alpha=.35, title=f'{x} vs {y}'); plt.show()
print('xG/xAG and passing/progression columns are not present in this source.')
```
